```python
import random

# the computer's notebook: how many times each move would have beaten the player's choice
move_scores = {"rock": 0, "scissors": 0, "paper": 0}

# beaten_by[x] is the move that beats x
beaten_by = {"rock": "paper", "scissors": "rock", "paper": "scissors"}
```

```python
player_score = 0
computer_score = 0

for round_number in range(15):
    player = input("Rock, scissors or paper? ").lower()
    if player not in beaten_by:
        print("Invalid input!")
        continue

    computer = max(move_scores, key=move_scores.get)
    if round_number < 3:
        computer = random.choice(["rock", "scissors", "paper"])
    print("Computer: ", computer)

    if computer == beaten_by[player]:
        print("Computer wins")
        computer_score = computer_score + 1
    elif computer == player:
        print("Draw")
    else:
        print("You win!")
        player_score = player_score + 1

    move_scores[beaten_by[player]] = move_scores[beaten_by[player]] + 1
    print("computer's notebook scores: ", move_scores)
    print()

print("You:", player_score, "      Computer:", computer_score)
```
